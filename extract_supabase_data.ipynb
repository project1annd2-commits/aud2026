# Supabase Data Extraction

This notebook will extract data from the Supabase database tables and save it for analysis. We'll extract data from:
- Schools
- Teachers
- Mentors
- Audits
- Infrastructure Audits

```python
# Import required libraries
import os
from dotenv import load_dotenv
import pandas as pd
from supabase import create_client, Client

# Load environment variables
load_dotenv()

# Initialize Supabase client
url = os.getenv('VITE_SUPABASE_URL')
key = os.getenv('VITE_SUPABASE_ANON_KEY')
supabase: Client = create_client(url, key)
```

```python
# Extract Schools data
schools_response = supabase.table('schools').select('*').execute()
schools_df = pd.DataFrame(schools_response.data)
print("Schools data shape:", schools_df.shape)
schools_df.head()
```

```python
# Extract Teachers data
teachers_response = supabase.table('teachers').select('*').execute()
teachers_df = pd.DataFrame(teachers_response.data)
print("Teachers data shape:", teachers_df.shape)
teachers_df.head()
```

```python
# Extract Mentors data
mentors_response = supabase.table('mentors').select('*').execute()
mentors_df = pd.DataFrame(mentors_response.data)
print("Mentors data shape:", mentors_df.shape)
mentors_df.head()
```

```python
# Extract Audits data
audits_response = supabase.table('audits').select('*').execute()
audits_df = pd.DataFrame(audits_response.data)
print("Audits data shape:", audits_df.shape)
audits_df.head()
```

```python
# Extract Infrastructure Audits data
infra_audits_response = supabase.table('infrastructure_audits').select('*').execute()
infra_audits_df = pd.DataFrame(infra_audits_response.data)
print("Infrastructure Audits data shape:", infra_audits_df.shape)
infra_audits_df.head()
```

```python
# Save all data to CSV files
output_dir = 'extracted_data'
os.makedirs(output_dir, exist_ok=True)

# Save each dataframe to CSV
schools_df.to_csv(f'{output_dir}/schools.csv', index=False)
teachers_df.to_csv(f'{output_dir}/teachers.csv', index=False)
mentors_df.to_csv(f'{output_dir}/mentors.csv', index=False)
audits_df.to_csv(f'{output_dir}/audits.csv', index=False)
infra_audits_df.to_csv(f'{output_dir}/infrastructure_audits.csv', index=False)

print("All data has been saved to CSV files in the 'extracted_data' directory")
```
